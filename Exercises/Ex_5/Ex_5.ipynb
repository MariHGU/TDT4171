<h1>Decision Support System</h1>

<h3>World description</h3>
<p>You are in the airport, and boarding for group A is just starting(you are group C). You have to pee, and is also hungry. The bathroom is right next to the kiosk, away from your gate. You cannot directly observe the bathroom line and the kiosk line from where you are. If the line to both are short/non-existent you can easily make them both before your group starts boarding. However if the line is long for either, the airplane is likely to complete boarding by the time you have returned. </br>If you choose not to go to the toilet, you  might have an embarrassing accident in the airplane, if you can not hold yourself until the toilets are available during the flight. You are also unsure of the food availability on the flight, meaning you may, or may not go hungry throughout the flight.</p>
<p>
    <b>Chance nodes</b> (Random Variables): bathroomLine(short, long), kioskLine(short, long), boardingSpeed(short, long), bladderStrength(weak, strong), inflightFood(available, not available)</br>
    <b>Decision nodes: </b>stay, go</br>
    <b>Utility nodes: </b>makingTheFlight(high utility), missingTheFlight(very low utility), hunger(low utility), bladderAccident(very negative utility), expensiveFood(low utility) 
</p>

<h3>Posterior possibilities and expected utilities of utility nodes:  </h3>
<p>NB! This is done under the assumption that bathroomLine is independent of kioskLine.</p>
<p>
    <b>Option 1: Stay at gate </b>=> P(makingFlight) = 1</br></br>
    P(Hunger) = P(inflighFood = unavailable) = 0.6, P(Accident) = P(bladderStrength = Weak) = 0.3</br>
    EU(stay) = ∑ P<sub>i</sub>*U<sub>i</sub> = P(makingFlight)*U(makingFlight) + P(accident)*U(accident) + P(hunger)*U(Hunger) = 1*100 + 0.3*(-50) + 0.6*(-20) = 73 </br></br>
</p>
<p>
    <b>Option 2: Go </b>=> P(makingFlight = false) = P(boardingSpeed = Fast)*P(bathroomLine = Long OR kioskLine = Long)</br></br>
    P(bathroomLine = Long AND kioskLine = Long) = 1 - P(bathroomLine = Short AND kioskLine = Short) = 1 - (0.6 * 0.5) = 0.7</br>
    => P(makingFlight = false) = 0.3*0.7 = 0.21 => P(makingFlight = true) = 1 - 0.21 = 0.79 </br>
    P(Accident) = 0, P(Hunger) = 0 </br>
    EU(go) = P(makingFlight)*U(makingFlight) + P(Hunger)*U(Hunger) + P(Accident)*U(Accident) = 100*0.79 + 0*0 + 0*0 = 79
</p>
<h3>Conclusion: </h3>
<h4>EU(stay) = 73 < EU(go) = 79 => A rational agent will choose to go to the bathroom and buy food at the kiosk.</h4>

</p>
<h3>Conclusion: </h3>
<h4>EU(stay) = 73 < EU(go) = 86.5 => A rational agent will choose to go to the bathroom and buy food at the kiosk.</h4>
