<h1>Task 2</h1>
<h3>Exp Utility function: U(x)= -<i>e<sup>-x/R</sup></i></h3>
<h3><i>a)</i></h3>
<h4>Base info: </h4>
<p>R = 500, U(loss) = U(0) = -1</br> Option 1: P(win) = 1.0 and U(500)= -0.368, Option 2: P(win) = 0.6 and U(5000)= -4.54 * 10<sup>-5</sup></p>
<p><b>MEU principle: </b>A rational agent must choose the action that maximizes expected utility. </br> EU(Option 1) <- P(win)*U(win) + P(Loss)*U(Loss) = 1*U(500) + 0*U(loss) = -0.368 </br>EU(Option 2) <- P(win)*U(win) + P(Loss)*U(Loss) = 0.6*U(5000) + 0.4*U(loss) = -0.4 </p>
<p>EU(Option 1) > EU(Option 2) => Assuming Mary is rational, should choose option 1.</p>

<h3><i>b)</i></h3>
<h4>Base info: </h4>
<p>R = ?, U(loss) = U(0) = -1 </br>Want indifference between option A and option B: EU(A)~EU(L<sub>B</sub>) | L<sub>B</sub> = [0.5, 500; 0.5, 0]</br></p>
<p>Equation to solve for R: -e<sup>-100/R</sup> = -e<sup>-500/R</sup>*0.5 + (-1)*0.5</br> Solved for R using scipy.optimize.fsolve()</p>

```python
import numpy as np
import scipy.optimize as so

def expected_utility_equation(R):
    # Certain amount utility (U(100))
    U_certain = -np.exp(-100 / R)

    # Lottery expected utility (0.5 * U(500) + 0.5 * U(0))
    U_lottery = 0.5 * (-np.exp(-500 / R)) + 0.5 * (-np.exp(0))

    # Indifference condition: EU(certain) = EU(lottery)
    return U_certain - U_lottery


R_initial_guess = 50
R_solution_corrected = so.fsolve(expected_utility_equation, R_initial_guess)[0]
R_solution_corrected_rounded = round(R_solution_corrected, 3)


print(f'Numerical solution to R: {R_solution_corrected_rounded}')
```
