<h3>Task 1:</h3>
<p>
    <b>a) </b> A Case is, in CBR technology, a problem situation. A case can be new, meaning unsolved, or past, meaning previously solved. The cases are stored in a problem-solution pair for easy accessibility.
</p>
<p>
    <b>b) </b>
    <ol>
        <li><b>Retrieve:</b> Previous knowledge is utilized in solving a new problem by retrieving a similar case to compare to the new problem.</li>
        <li><b>Reuse:</b> By reusing the knowledge learned from the previous case, we can create a suggested solution to our problem.</li>
        <li><b>Revise:</b> By testing the initial solution in a real life enviornment we can revise the solution, meaning to improve the solution of our problem.</li>
        <li><b>Retain:</b> After finding a final solution we can retain the valuable information we learned for solving future cases.</li>
    </ol>
</p>
